```python
# ord(char) gets the ASCII value of the character (e.g., '1' is 49).

# ord('0') is 48, so ord('1') - ord('0') = 49 - 48 = 1.
```

```python
def string_to_int(s):
    result = 0
    for char in s:
        # Convert character to digit by subtracting ASCII value of '0' (48)
        digit = ord(char) - ord('0')
        print(digit)
        # Shift current result left (multiply by 10) and add new digit
        result = result * 10 + digit
        print(result)
    return result
```

```python
test_string = "123"
number = string_to_int(test_string)
print(number)  # Output: 123
```
