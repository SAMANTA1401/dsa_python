```python
def searchSortedMatrix(matrix,target):
    # number of rows
    m = len(matrix) # 3
    if m == 0:
        return False
    ## number of columns 
    n =len(matrix[0]) # 4
    # Binary search implementation
    left, right = 0,m*n-1 # 0,11
    while left <= right:
        mid = left + (right - left)//2  #5
        print(mid)
        # row [[1,3,5,7],[10,11,16,20],[23,30,34,60]]
        mid_element = matrix[mid//n][mid%n] # 5//4, 5%4 = 1,2 >> 11
        print(mid_element)
        if target == mid_element:
            print(mid//n,mid%n)
            return True
        elif target < mid_element:
            right = mid - 1
        else:
            left = mid + 1
    return False

# time complexity O(log(m*n)) total number of elements in the matrix)
# space complexity O(1) uses a constant amount of space to store the variables:

```

```python
matrix=[[1,3,5,7],[10,11,16,20],[23,30,34,60]]
target= 10
# print(len(matrix))
# print(len(matrix[0]))
result = searchSortedMatrix(matrix, target)
print(result)
```
